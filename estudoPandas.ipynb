# Problema 2

- groupby: https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.groupby.html
- Grouper: https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Grouper.html

```python
from datetime import datetime, timedelta
base = datetime.today()
base_times = 900
date_list = [base - timedelta(days=x) for x in range(base_times)] # A partir da data atual esta criando uma lista com 900 dias no passado
```

```python
date_list
```

```python
freq = Series(
    [randint(1, 10_000) for _ in range(base_times)],
    index=date_list
)
```

```python
freq
```

```python
freq.iloc[0:30]
```

```python
from pandas import Grouper

g = Grouper(freq='1YE')
```

```python
m = Grouper(freq='1ME')
```

```python
freq.groupby(m).count()
```

```python
freq.iloc[0:30].groupby(Grouper(freq='1W')).sum().plot(kind='bar')
```
